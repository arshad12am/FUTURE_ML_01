```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
```

```python
numerical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)
```

```python
categorical_transformer = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ]
)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numerical_transformer, numerical_columns),
        ("cat", categorical_transformer, categorical_columns)
    ]
)
```

```python
def calculate_test_lag_features(row):
    store = row["Store"]
    current_date = row["Date"]

    # Previous day
    lag_1 = historical_sales.get(
        (store, current_date - pd.Timedelta(days=1)),
        np.nan
    )

    # Seven days earlier
    lag_7 = historical_sales.get(
        (store, current_date - pd.Timedelta(days=7)),
        np.nan
    )

    # Previous 7 calendar days
    previous_seven_days = historical_sales.reindex(
        pd.MultiIndex.from_product(
            [
                [store],
                pd.date_range(
                    current_date - pd.Timedelta(days=7),
                    current_date - pd.Timedelta(days=1)
                )
            ],
            names=["Store", "Date"]
        )
    )

    rolling_7 = previous_seven_days.mean()

    return pd.Series({
        "Sales_Lag_1": lag_1,
        "Sales_Lag_7": lag_7,
        "Sales_Rolling_7": rolling_7
    })


test_features[
    ["Sales_Lag_1", "Sales_Lag_7", "Sales_Rolling_7"]
] = test_features.apply(
    calculate_test_lag_features,
    axis=1
)
```

```python
import numpy as np
import pandas as pd

def add_competition_features(df):
    df = df.copy()

    # Treat year 1900 as an invalid/unknown placeholder
    valid_competition_date = (
        df["CompetitionOpenSinceYear"].notna()
        & df["CompetitionOpenSinceMonth"].notna()
        & (df["CompetitionOpenSinceYear"] != 1900)
    )

    # Indicator showing whether the competition start date is known
    df["CompetitionStartKnown"] = valid_competition_date.astype(int)

    # Build the competition opening date only where it is valid
    competition_start = pd.to_datetime(
        {
            "year": df["CompetitionOpenSinceYear"].where(
                valid_competition_date
            ),
            "month": df["CompetitionOpenSinceMonth"].where(
                valid_competition_date
            ),
            "day": 1
        },
        errors="coerce"
    )

    # Competition is active when its opening date is before or equal to
    # the current observation date
    df["CompetitionActive"] = (
        valid_competition_date
        & (df["Date"] >= competition_start)
    ).astype(int)

    # Number of months since competition opened
    df["MonthsSinceCompetitionOpened"] = np.nan

    active_mask = (
        valid_competition_date
        & (df["Date"] >= competition_start)
    )

    df.loc[active_mask, "MonthsSinceCompetitionOpened"] = (
        (df.loc[active_mask, "Date"].dt.year
         - competition_start.loc[active_mask].dt.year) * 12
        +
        (df.loc[active_mask, "Date"].dt.month
         - competition_start.loc[active_mask].dt.month)
    )

    return df
```

```python
# ============================================================
# Random Forest — 30% Sample + Expanding Time-Series CV
# ============================================================

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

# ------------------------------------------------------------
# 1. Features
# ------------------------------------------------------------

best_lag_features = [
    "Sales_Lag_7",
    "Sales_Lag_14",
    "Sales_Rolling_14"
]

selected_features = feature_columns + best_lag_features

numeric_features = [
    "Store",
    "DayOfWeek",
    "Open",
    "Promo",
    "SchoolHoliday",
    "CompetitionDistance",
    "CompetitionOpenSinceMonth",
    "CompetitionOpenSinceYear",
    "Promo2",
    "Promo2SinceWeek",
    "Promo2SinceYear",
    "Year",
    "Month",
    "Day",
    "WeekOfYear"
] + best_lag_features

categorical_features = [
    "StateHoliday",
    "StoreType",
    "Assortment",
    "PromoInterval"
]

# ------------------------------------------------------------
# 2. Use only the original training period
# ------------------------------------------------------------

cv_data = train_features.loc[train_mask].copy()

# Sort chronologically
cv_data = cv_data.sort_values("Date").reset_index(drop=True)

# ------------------------------------------------------------
# 3. Create 3 expanding temporal folds
# ------------------------------------------------------------

fold_dates = [
    pd.Timestamp("2014-01-01"),
    pd.Timestamp("2014-10-01"),
    pd.Timestamp("2015-04-01")
]

results = []

for fold, validation_start in enumerate(fold_dates, start=1):

    validation_end = validation_start + pd.DateOffset(months=3)

    fold_train = cv_data[
        cv_data["Date"] < validation_start
    ]

    fold_valid = cv_data[
        (cv_data["Date"] >= validation_start) &
        (cv_data["Date"] < validation_end)
    ]

    # Sample only the training portion
    fold_train_sample = fold_train.sample(
        frac=0.30,
        random_state=42
    )

    X_train = fold_train_sample[selected_features]
    y_train = fold_train_sample["Sales"]

    X_valid = fold_valid[selected_features]
    y_valid = fold_valid["Sales"]

    # --------------------------------------------------------
    # 4. Preprocessing
    # --------------------------------------------------------

    numeric_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ])

    categorical_transformer = Pipeline(steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        ("onehot", OneHotEncoder(handle_unknown="ignore"))
    ])

    preprocessor = ColumnTransformer(
        transformers=[
            ("num", numeric_transformer, numeric_features),
            ("cat", categorical_transformer, categorical_features)
        ]
    )

    # --------------------------------------------------------
    # 5. Random Forest
    # --------------------------------------------------------

    rf_model = RandomForestRegressor(
        n_estimators=100,
        max_depth=20,
        min_samples_leaf=10,
        random_state=42,
        n_jobs=-1
    )

    rf_pipeline = Pipeline(steps=[
        ("preprocessor", preprocessor),
        ("model", rf_model)
    ])

    # --------------------------------------------------------
    # 6. Train and evaluate
    # --------------------------------------------------------

    print(
        f"\nFold {fold}: "
        f"Train={len(fold_train_sample):,}, "
        f"Validation={len(fold_valid):,}"
    )

    rf_pipeline.fit(X_train, y_train)

    predictions = rf_pipeline.predict(X_valid)

    mae = mean_absolute_error(y_valid, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_valid, predictions)
    )

    r2 = r2_score(y_valid, predictions)

    results.append({
        "Fold": fold,
        "Train_Rows": len(fold_train_sample),
        "Valid_Rows": len(fold_valid),
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    })

    print(f"MAE:  {mae:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"R²:   {r2:.4f}")

# ------------------------------------------------------------
# 7. Summary
# ------------------------------------------------------------

cv_results = pd.DataFrame(results)

print("\nFold Results")
print(cv_results.round(4))

print("\n30% Sample Time-Series CV Summary")
print("----------------------------------")
print(f"Average MAE:  {cv_results['MAE'].mean():.2f}")
print(f"Average RMSE: {cv_results['RMSE'].mean():.2f}")
print(f"Average R²:   {cv_results['R2'].mean():.4f}")
print(f"R² Std:       {cv_results['R2'].std():.4f}")
```

```python
def evaluate_lr(feature_list):
    selected_categorical = [
        feature for feature in feature_list
        if feature in categorical_features
    ]

    selected_numeric = [
        feature for feature in feature_list
        if feature not in categorical_features
    ]

    transformers = []

    if selected_numeric:
        numeric_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="median"))
        ])

        transformers.append(
            ("numeric", numeric_pipeline, selected_numeric)
        )

    if selected_categorical:
        categorical_pipeline = Pipeline([
            ("imputer", SimpleImputer(strategy="most_frequent")),
            ("encoder", OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ))
        ])

        transformers.append(
            ("categorical", categorical_pipeline, selected_categorical)
        )

    preprocessor = ColumnTransformer(transformers)

    model = Pipeline([
        ("preprocessor", preprocessor),
        ("model", LinearRegression())
    ])

    start_time = time.time()

    model.fit(
        X_train_full[feature_list],
        y_train
    )

    predictions = model.predict(
        X_valid_full[feature_list]
    )

    elapsed_time = time.time() - start_time

    return {
        "Features": feature_list.copy(),
        "Feature_Count": len(feature_list),
        "MAE": mean_absolute_error(y_valid, predictions),
        "RMSE": np.sqrt(mean_squared_error(y_valid, predictions)),
        "R2": r2_score(y_valid, predictions),
        "Time": elapsed_time
    }
```

```python
import time
import numpy as np
import pandas as pd

from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

# --------------------------------------------------
# Features
# --------------------------------------------------

rf_features = [
    "Open",
    "Sales_Lag_14",
    "Sales_Lag_7",
    "Sales_Rolling_14",
    "Sales_Rolling_7",
    "Day",
    "Promo"
]

X = train_features[rf_features].copy()
y = train_features["Sales"].copy()

# --------------------------------------------------
# Five chronological folds
# --------------------------------------------------

n_samples = len(X)
fold_size = n_samples // 5
total_folds = 5

results = []

for fold in range(total_folds):

    # First fold: train on the first 20%, validate on the second 20%
    train_start = 0
    train_end = fold_size * (fold + 1)

    valid_start = train_end
    valid_end = (
        fold_size * (fold + 2)
        if fold < total_folds - 1
        else n_samples
    )

    X_train = X.iloc[train_start:train_end]
    y_train = y.iloc[train_start:train_end]

    X_valid = X.iloc[valid_start:valid_end]
    y_valid = y.iloc[valid_start:valid_end]

    print("\n" + "=" * 70)
    print(f"STARTING FOLD {fold + 1}/{total_folds}")
    print(f"Training rows:   {len(X_train):,}")
    print(f"Validation rows: {len(X_valid):,}")
    print("=" * 70)

    model = RandomForestRegressor(
        n_estimators=100,
        max_features="sqrt",
        random_state=42,
        n_jobs=-1
    )

    start_time = time.time()

    model.fit(X_train, y_train)
    predictions = model.predict(X_valid)

    elapsed_time = time.time() - start_time

    mae = mean_absolute_error(y_valid, predictions)
    rmse = np.sqrt(mean_squared_error(y_valid, predictions))
    r2 = r2_score(y_valid, predictions)

    results.append({
        "Fold": fold + 1,
        "Train_Size": len(X_train),
        "Validation_Size": len(X_valid),
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2,
        "Time_sec": elapsed_time
    })

    progress = ((fold + 1) / total_folds) * 100

    print("\n" + "-" * 70)
    print(f"FOLD {fold + 1} COMPLETED")
    print(f"MAE:       {mae:.6f}")
    print(f"RMSE:      {rmse:.6f}")
    print(f"R²:        {r2:.6f}")
    print(f"Time:      {elapsed_time:.2f} seconds")
    print(f"PROGRESS:  {fold + 1}/{total_folds} folds completed ({progress:.0f}%)")
    print("-" * 70)

    # Display all completed fold results so far
    display(pd.DataFrame(results))

# --------------------------------------------------
# Final summary
# --------------------------------------------------

rf_full_results = pd.DataFrame(results)

print("\n" + "=" * 70)
print("FINAL RESULTS — 5 FOLDS")
print("=" * 70)

display(rf_full_results)

print(f"Average MAE:  {rf_full_results['MAE'].mean():.6f}")
print(f"Average RMSE: {rf_full_results['RMSE'].mean():.6f}")
print(f"Average R²:   {rf_full_results['R2'].mean():.6f}")
print(f"Average Time: {rf_full_results['Time_sec'].mean():.6f} sec")
```
